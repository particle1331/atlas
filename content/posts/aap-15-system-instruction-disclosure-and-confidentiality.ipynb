This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Dummy markers, direct extraction and code-framed extraction.
- Controlled distractor length and repeated trials, without assuming that a short prompt fills the context window.
- Exact disclosure, partial disclosure and refusal as measured outcomes.
- Why system-instruction secrecy is not an authorization boundary and real credentials do not belong in such prompts.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap15` (relative to the repository root).

- A dummy-marker prompt fixture varies extraction framing and distractor lengths under recorded model settings.
- A scorer distinguishes exact disclosure, partial disclosure, refusal, and unrelated output.
- Repeated conditions test behaviour without using real secrets or asserting an unmeasured attention mechanism.
