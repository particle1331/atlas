## Topics

- Deterministic success, timeout and unavailable-service injection.
- API failure versus tool failure; error disclosure versus recovery.
- Retry budgets, backoff assumptions and idempotency through a simulated effect ledger.
- Partial results, exhausted retries and fresh output from the adapted code.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap11` (relative to the repository root).

- A deterministic weather/service fixture injects success, timeout, and unavailable-service results.
- A retry controller and simulated side-effect ledger separate error reporting from recovery and duplicated work.
- Scenario tests compare partial results, exhausted budgets, and disclosure without relying on model-generated failures.
