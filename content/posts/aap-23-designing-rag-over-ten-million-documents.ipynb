This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Corpus/chunk sizes, change rate, query load, authorization and latency requirements.
- Ingestion, index build/update, partitioning, retrieval, reranking and grounded answer interfaces.
- Storage/embedding/index/update capacity estimates and identified bottlenecks.
- Bounded integration/workload simulation, failure cases and cache invalidation; separate estimates from measured scale.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap23` (relative to the repository root).

- A bounded ingestion/index/query slice and workload generator exercise the proposed large-corpus interfaces.
- A capacity-estimation model states document/chunk sizes, update rates, query load, and storage/index assumptions.
- Integration tests cover freshness, authorization, retrieval quality, partial failures, and cache invalidation without claiming a ten-million-document run.

**Starting material:** Source design exercise 2/7; AAP-16–22.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.

## Known gaps and limitations

The following issues remain unresolved; no implementation or experiment is claimed to have closed them.

| Gap | Current limitation | Evidence still needed |
|---|---|---|
| G14 | Metadata retrieval is hand-selected, security corpus unused; extracted filters are not authorization or a zero-hallucination guarantee. | Implement real retrieval/filtering, trusted access and partial-filter semantics; narrow guarantees. |
| G17 | Correct middle-context answer is labelled failure via ranking precision; empty `supported` list yields faithfulness 1.0. | Separate retrieval/generation metrics, control position/length, and validate judge outputs/abstention. |
| G31 | Semantic caching has no adequate audited implementation; prompt caching is a different mechanism. | Build semantic reuse with equivalence/freshness/access checks and wrong-reuse/verification-cost measures. |
| G34 | Seven design labels are not seven demonstrated full implementations; toy runs do not establish scale/reliability. | State workload/trust/failure assumptions, implement a bounded slice/simulation and separate results from estimates/unbuilt architecture. |
