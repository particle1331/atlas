This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Fact identity, ADD/UPSERT/DELETE actions and schema/domain validation.
- Duplicates, contradictions, provenance, deletion and missing/unknown actions.
- Deterministic update tests separate from extraction/action-selection quality.
- Persistence added to the volatile example where needed; evaluate stale and incorrect updates.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap31` (relative to the repository root).

**Imported reference examples:** `reference/evolving-memory/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- Validated ADD/UPSERT/DELETE proposals feed a deterministic mutation engine with explicit fact identity and provenance.
- Durable writes and change reporting are separate from model action selection.
- Unit cases cover duplicates, contradictions, corrections, deletion, missing actions, and invalid shapes; model extraction is evaluated independently.

**Starting material:** `evolving-memory`, extending AAP-30 rather than treating mutation as persistence.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
