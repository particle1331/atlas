This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Baseline answers, persuasive pushback and multi-turn state.
- Matched pressure conditions: confidence, authority, emotional appeal and genuine correction.
- Ambiguous questions, restrictive output formats and the limits of exact answer persistence as a metric.
- Repeated trials scoring factual accuracy and warranted updates, not stubbornness as a proxy for quality.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap6` (relative to the repository root).

- A conversation builder creates matched baseline and pushback histories without changing the underlying question.
- Pressure variants and evidence-bearing corrections are labelled separately; the scorer checks warranted updates rather than persistence alone.
- Repeated trials compare accuracy and justified revisions across the same factual and ambiguous cases.

**Starting material:** `sycophancy`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
