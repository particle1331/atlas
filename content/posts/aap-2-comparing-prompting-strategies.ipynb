This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Direct instructions, requested intermediate explanations and worked examples as different experimental conditions.
- Parameterized tasks and reliable final-answer checks beyond familiar puzzles.
- Repeated matched comparisons, per-case outcomes and sample uncertainty.
- Example selection/count and input/output overhead; generated explanations are not proof of internal reasoning fidelity.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap2` (relative to the repository root).

**Imported reference examples:** `reference/cot-vs-direct/`, `reference/cot-vs-few-vs-direct/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A parameterized problem generator supplies held-out cases and an independent final-answer oracle.
- Prompt builders define direct, step-by-step, and few-shot conditions while a shared runner records paired outcomes and token usage.
- The experiment varies examples and task difficulty without treating generated explanations as an observable internal reasoning trace.

**Starting material:** `cot-vs-direct`, `cot-vs-few-vs-direct`, treated as a paired comparison rather than duplicate lessons.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
