This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- JSON requested in prose versus provider-supported response schemas.
- Parsing, types, enums, cross-field constraints and semantic correctness.
- Canonicalized object comparison versus exact-string equality.
- Malformed/refused/truncated outputs and schema-valid wrong answers as independent test cases.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap3` (relative to the repository root).

**Imported reference examples:** `reference/non-deterministic/`, `reference/evals/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A shared request fixture is sent through free-form, JSON-instruction, and provider response-schema adapters.
- Parsing, schema validation, domain checks, and answer checks are separate functions with independent failure records.
- The test cases include schema-valid wrong answers, refusals, truncation, and equivalent objects with different serialization.

**Starting material:** `non-deterministic`, structured examples in `evals`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
