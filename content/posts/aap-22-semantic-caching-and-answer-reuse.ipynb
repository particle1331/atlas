This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Exact response caching, semantic answer reuse and provider prompt/context caching as different mechanisms.
- Context-sensitive cache keys: tenant, authorization, source version, time, task and configuration.
- Similarity proposes a candidate; it does not establish answer equivalence.
- Negation/date/unit counterexamples; false reuse, missed reuse, verification cost, freshness and latency.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap22` (relative to the repository root).

- A bounded response store keys entries by task, tenant/access scope, source version, time dependence, and configuration.
- Similarity lookup proposes candidates while an explicit equivalence/freshness policy decides whether reuse is permitted.
- Counterexamples involving negation, dates, and units measure false reuse, missed reuse, verification cost, and latency.

**Starting material:** Source semantic-caching coverage; new bounded implementation, not renamed `prompt-caching`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
