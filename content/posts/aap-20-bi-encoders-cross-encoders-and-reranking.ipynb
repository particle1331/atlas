This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Independent embeddings versus joint query/document scoring.
- A genuine candidate-generation→reranking boundary with fixed candidate sets.
- Ambiguity, negation and directional relevance cases with explicit judgements.
- Candidate-depth recall, final ranking and latency; scores are not calibrated probabilities without evidence.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap20` (relative to the repository root).

**Imported reference examples:** `reference/cross-encoders/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A bi-encoder candidate generator passes a fixed candidate set to a cross-encoder pair scorer.
- Relevance fixtures cover ambiguity, negation, and directional queries with independently stated judgements.
- Candidate-depth recall, final ranking, and latency are compared without interpreting scores as calibrated probabilities.
