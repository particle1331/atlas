This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Rubric and free-form feedback on a shared initial design.
- Iterative edits, stopping criteria and cost of additional critique rounds.
- Independent concrete defect checks or blinded assessment; ceiling effects and judge agreement.
- Cases where scores rise without better work, or useful feedback does not raise the score.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap7` (relative to the repository root).

**Imported reference examples:** `reference/critic-refiner/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A shared initial design feeds rubric-guided and free-form critique/refinement loops.
- An independent defect checklist and optional blinded assessment are kept outside the critic's own scoring path.
- The comparison records edits, residual defects, judge ratings, round limits, and costs to test whether score changes track improvements.
