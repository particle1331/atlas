This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Conversation state versus external persistence and scope.
- Extraction, stored representation, provenance and runtime/sample fixtures.
- Keyword matching, paraphrases and selection into later context.
- Controlled persistence/read tests and the limits of append-only memory.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap30` (relative to the repository root).

**Imported reference examples:** `reference/memory-persistent/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A fact extractor, persistent store, and memory-read interface separate conversation state from externally retained facts.
- Fixtures record fact provenance, storage scope, and keyword/paraphrase retrieval results.
- Cross-session tests measure useful recall, missed recall, stale facts, and the limits of append-only storage.

**Starting material:** `memory-persistent`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
