This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Conversation state versus external persistence and scope.
- Extraction, stored representation, provenance and runtime/sample fixtures.
- Keyword matching, paraphrases and selection into later context.
- Controlled persistence/read tests and the limits of append-only memory.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap30` (relative to the repository root).

- A fact extractor, persistent store, and memory-read interface separate conversation state from externally retained facts.
- Fixtures record fact provenance, storage scope, and keyword/paraphrase retrieval results.
- Cross-session tests measure useful recall, missed recall, stale facts, and the limits of append-only storage.

**Starting material:** `memory-persistent`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
