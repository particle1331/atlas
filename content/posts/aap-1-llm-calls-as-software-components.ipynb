This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Request/response boundaries and the division of responsibility between model and application.
- Sampling variation, refusal, truncation, malformed output, missing information and service errors.
- Repeated extraction with separate formatting, parsing and correctness checks.
- Task baselines, configuration and error accounting; what repeated output does not prove.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap1` (relative to the repository root).

- A small extraction dataset, a configurable request adapter, and a repeated-call runner separate task inputs from provider configuration.
- A result recorder distinguishes service errors, parse failures, valid outputs, and correct answers; no error string is counted as an ordinary answer.
- The comparison reports exact-text variation, canonicalized values, correctness, latency, and usage for the same cases.

**Starting material:** `non-deterministic`; selected classification examples from `evals`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
