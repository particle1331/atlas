This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Patch proposals, application context and actual changed lines.
- Malformed/stale patches, partial application and failed fallback.
- Disposable worktrees, unrelated staged/unstaged work and acceptance before staging.
- Tests and rollback; compare edit surface/reviewability with semantic correctness.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap27` (relative to the repository root).

**Imported reference examples:** `reference/code-patcher/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A disposable worktree receives proposed unified diffs through a validated patch-application interface.
- Change inspection, behavioural tests, acceptance, staging, and rollback are distinct operations.
- Cases cover malformed/stale/partial patches, fallback failure, and unrelated staged changes, compared with whole-file repair.

**Starting material:** `code-patcher`, compared with the AAP-24 whole-file approach.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
