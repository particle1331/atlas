This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Proposed action, operator-visible evidence and review decisions.
- Operation identity, arguments, environment, expiry and one-shot approval.
- Mocked denial, approval replay, changed context and failure after approval.
- Host-shell/Python permissions; a timeout or a prompt is not a sandbox.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap12` (relative to the repository root).

**Imported reference examples:** `reference/hitl/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- An operation proposal and one-shot approval store bind the decision to arguments, execution context, expiry, and identity.
- A mocked executor enforces approval in application code instead of relying on a prompt or terminal confirmation.
- Tests exercise denial, changed context, repeated approval use, expiration, and failure after approval; any real execution requires containment.

**Starting material:** `hitl`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
