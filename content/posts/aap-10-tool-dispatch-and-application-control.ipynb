This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Parsing, validation, tool registries and operation/result correlation.
- Trusted permissions, resource scopes, timeouts and structured results.
- Multiple requested calls and application-chosen sequential/concurrent execution.
- Offline tests for unknown tools, extra fields, invalid values and protocol preservation.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap10` (relative to the repository root).

**Imported reference examples:** `reference/loop/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A registry-backed dispatcher validates requests before invoking fixture tools under explicit permission and resource rules.
- Operation identifiers, structured results, timeouts, and provider message adapters preserve request/result correlation.
- Offline tests cover unknown tools, extra fields, invalid values, multiple calls, and sequential versus concurrent execution.

**Starting material:** Dispatcher excerpts from `loop`; tool interfaces developed in AAP-9.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
