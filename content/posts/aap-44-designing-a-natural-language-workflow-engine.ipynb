## External resources

Arpit Bhayani's incident-triage experiment offers a narrower comparison:

- [Compiling a Workflow into a 3B Model](https://arpitbhayani.me/blogs/compiling-an-sop-into-a-3b-model/) — the experiment's write-up.
- [workflow-slm](https://github.com/arpitbbhayani/workflow-slm) — companion code and documentation for fine-tuning a small model to follow a predefined standard operating procedure (SOP).

The repository explores moving the procedure from the prompt into model weights. At each turn, the model proposes a diagnostic action; the runtime validates its schema and transition order before calling simulated tools. Deterministic code computes severity and paging decisions. This separates learned action selection from explicit execution rules.

Following a fixed SOP differs from translating an arbitrary request into a new workflow. The example is useful for studying bounded tool use and failure handling, but does not establish the broader engine's authorization, durable execution or compensation mechanisms. It is an external reference, not the missing course implementation; its reported results have not been reproduced here.
