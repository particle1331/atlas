This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Constrained workflow representation, ambiguity resolution and proposal versus authorization.
- Tool registry, typed dependencies, scheduling and durable state.
- Approval, retries, cancellation, compensation and observable execution.
- End-to-end malformed-plan, denied-action, partial-failure and resource-limit tests; when a conventional workflow is enough.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap44` (relative to the repository root).

- A constrained workflow representation separates natural-language proposals, ambiguity resolution, authorization, and execution.
- A tool registry, dependency scheduler, durable state, review points, and compensation/cancellation interfaces compose earlier mechanisms.
- Integration cases cover malformed plans, denied actions, partial failure, and exhausted budgets and compare with an ordinary non-model workflow.

**Starting material:** Source design exercise 7/7; integration of dispatch, state, coordination, evaluation and operation.
