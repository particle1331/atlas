This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Agent anatomy: model, harness, state, tools, observations and stopping policy.
- Observe–act/ReAct representations and preservation of required provider message state.
- Fragmented/interleaved tool-call argument assembly, completion and validation before dispatch, and interrupted streams tested with deterministic synthetic inputs.
- Multiple calls, result correlation and application-selected scheduling.
- No-tool completion, unknown calls, repeated errors, turn exhaustion and honest final-answer validation.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap25` (relative to the repository root).

- A streaming accumulator assembles fragmented/interleaved tool-call arguments and an explicit state machine preserves provider message requirements.
- Call completion and argument validation precede dispatch; scheduling, correlation, termination, and error budgets remain application-owned.
- Deterministic synthetic streams test interruption, multiple calls, no-tool completion, unknown tools, repeated errors, and turn exhaustion before model-dependent trials.

**Starting material:** `loop`, `react`, compared as iterative harness implementations rather than repeating AAP-10's single dispatch.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
