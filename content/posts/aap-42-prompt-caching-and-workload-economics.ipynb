This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Explicit/implicit provider caching under verified current rules, distinct from AAP-22 answer reuse.
- Stable prefixes, changed content, TTL, observed hits and cleanup.
- Verified usage/prices and applicable creation/storage costs.
- Actual cached/uncached comparisons and break-even conditions, including workloads where caching does not help.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap42` (relative to the repository root).

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A provider cache adapter records stable prefixes, changed content, TTL, creation/deletion, and actual hit metadata.
- A workload runner and current-price cost model separate observed usage from estimated storage/creation costs.
- Cached and uncached conditions include break-even and no-benefit workloads, without substituting answer reuse for provider context caching.
