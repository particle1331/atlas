This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Diff/repository scope, invariants and evidence required for a finding.
- Actual static tools/tests plus targeted model analysis, not reports from repository names.
- Structured findings, line evidence, severity, uncertainty and duplicate handling.
- Seeded-defect evaluation, false positives, missed defects, hostile source comments and isolated patch validation.

- Finding records require inspected line evidence, severity reasoning, uncertainty, and duplicate handling.
- Seeded defects and benign changes measure useful findings, false positives, misses, hostile comments, and isolated patch validation.
