This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Trusted telemetry, service ownership, incident/task model and permitted actions.
- Diagnosis/proposal specialists separated from an authorized executor.
- Review thresholds, operation identity, duplicate events, rollback/compensation and escalation.
- Bounded simulation of false diagnosis, stalled specialists and interrupted remediation; no real infrastructure changes.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap37` (relative to the repository root).

- An incident simulator provides trusted telemetry, task ownership, and allowed remediation operations.
- Diagnosis/proposal workers are separated from approval and an effect executor with operation identity and rollback/compensation records.
- End-to-end cases cover false diagnosis, duplicate events, stalled workers, interrupted remediation, and escalation without touching real infrastructure.
