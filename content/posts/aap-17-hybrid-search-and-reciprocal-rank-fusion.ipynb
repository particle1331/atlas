This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Actual lexical and learned-embedding baselines beside a labelled toy concept-space example.
- Rank rather than raw-score combination; derive smoothing effects and missing-list contributions.
- Candidate depth, absent documents, correlated errors and complementary retrieval.
- Labelled evaluation across smoothing values; interactive curves support reasoning rather than prove a universal default.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap17` (relative to the repository root).

**Imported reference examples:** `reference/rrf/`, `reference/rrf-viz/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- Lexical and learned-embedding retrievers produce independent candidate lists beside a labelled toy concept-space example.
- A small reciprocal-rank-fusion function handles smoothing, list truncation, and missing candidates explicitly.
- A labelled query benchmark and optional visualization compare complementary and correlated rankers rather than assume a universal smoothing constant.

**Starting material:** `rrf`, `rrf-viz`, paired because the visualization illuminates the same mechanism.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
