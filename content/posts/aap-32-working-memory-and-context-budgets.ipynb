This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Instructions, messages, tool schemas/results and retained state in actual token accounting.
- Derive cumulative history-resend cost under stated payload assumptions.
- Compare eviction, summaries and retrieval on one fact/evidence set.
- Rare but important detail loss, retrieval/embedding failures, task success, latency and cost; this is not provider caching.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap32` (relative to the repository root).

**Imported reference examples:** `reference/working-memory/`, `reference/react/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A shared conversation/fact fixture feeds eviction, summarization, and retrieval-assisted context builders.
- Usage accounting includes relevant instructions, messages, tools, and results and surfaces embedding failures.
- The comparison measures retained evidence, task correctness, latency, and cost under the same total-input budget.

**Starting material:** `working-memory`; accounting in `react`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
