This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Instructions, messages, tool schemas/results and retained state in actual token accounting.
- Derive cumulative history-resend cost under stated payload assumptions.
- Compare eviction, summaries and retrieval on one fact/evidence set.
- Rare but important detail loss, retrieval/embedding failures, task success, latency and cost; this is not provider caching.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap32` (relative to the repository root).

- A shared conversation/fact fixture feeds eviction, summarization, and retrieval-assisted context builders.
- Usage accounting includes relevant instructions, messages, tools, and results and surfaces embedding failures.
- The comparison measures retained evidence, task correctness, latency, and cost under the same total-input budget.

**Starting material:** `working-memory`; accounting in `react`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
