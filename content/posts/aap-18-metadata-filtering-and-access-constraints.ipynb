This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Real retrieval under a candidate limit instead of a hand-selected naive result.
- Extracted year/quarter constraints, ambiguous queries and partial/missing filters.
- Pre-filter versus post-filter effects on recall and candidate exhaustion.
- Authorization from trusted identity/policy, independently of model-extracted preferences.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap18` (relative to the repository root).

- A real retrieval fixture attaches metadata and separates model-extracted query filters from trusted authorization constraints.
- Pre-filter and post-filter adapters expose candidate limits, missing filters, and partial constraints.
- The same labelled cases measure recall, candidate exhaustion, invalid filters, and forbidden-document exclusion.

**Starting material:** `metadata-filtering`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
