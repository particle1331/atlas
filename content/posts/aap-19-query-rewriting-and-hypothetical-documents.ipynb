This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Vocabulary mismatch, ambiguity and hypothesized evidence text.
- Original-query embedding, query-to-query rewriting, and generation then embedding of a hypothetical document to retrieve real documents.
- A clearly labelled handwritten diagnostic condition; synthetic text is a retrieval aid, not factual evidence.
- Repeated transformations and harmful expansions, not only a selected rank flip.
- Candidate recall, ranking, latency and generation cost before any reranker is introduced.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap19` (relative to the repository root).

- Query builders distinguish original-query embedding, query-to-query rewriting, and generation then embedding of a hypothetical document, beside a labelled handwritten diagnostic condition.
- A common embedding/index runner retrieves real documents and records transformation provenance; synthetic text is not treated as factual evidence.
- The evaluation includes harmful expansions, ambiguity, recall/rank changes, latency, and generation cost.

**Starting material:** `query-rewrite-hyde`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
