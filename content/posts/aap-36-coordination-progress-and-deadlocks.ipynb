This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Distinguish long work, repeated routing, livelock and blocked wait-for deadlock.
- Reproduce circular refund/shipping prerequisites and the missing transition.
- State/visit records, progress measures, repeated-state checks and limits.
- Revised prerequisites, routing, cancellation and escalation; inject slow/failed workers.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap36` (relative to the repository root).

- A small routing simulator records task state, agent visits, prerequisites, and externally visible progress.
- Circular refund/shipping rules reproduce repeated handoffs alongside long-valid-work and blocked-resource controls.
- Tests compare progress/repeated-state checks, limits, cancellation, and changed prerequisites instead of equating every long run with deadlock.

**Starting material:** `deadlock`; coordination paths in AAP-35.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
