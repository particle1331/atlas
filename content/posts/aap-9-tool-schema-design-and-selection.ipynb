This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Distinguish names/descriptions from parameter structures; vary one contribution at a time.
- Selection, argument validity, domain validity and no-tool decisions as separate outcomes.
- Ambiguous queries, overlapping tools and underspecified requests.
- Repeated comparisons with per-case records instead of an arbitrary “significant improvement” threshold.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap9` (relative to the repository root).

- A tool-declaration registry exposes controlled naming, description, and parameter-schema variants.
- A labelled query suite checks selection, no-tool decisions, argument shape, and domain correctness separately.
- The comparison changes one interface contribution at a time and retains ambiguous cases and per-trial outcomes.

**Starting material:** `tool-schema-calibration`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
