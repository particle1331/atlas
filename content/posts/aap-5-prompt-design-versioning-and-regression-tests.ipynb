This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Drafting prompts from task definitions, evidence requirements, examples and consumer contracts.
- Prompt/configuration versions, dataset versions, recorded responses and a regression harness.
- A concrete change that fixes one case and breaks another; inspect the actual prompt diff.
- Held-out/per-class cases, repeated calls, compatibility and rollback.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap5` (relative to the repository root).

**Imported reference examples:** `reference/prompt-regression/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A versioned prompt registry and labelled classification fixtures supply inputs to a common regression runner.
- Per-case differences, prompt/configuration versions, errors, and repeated responses are retained rather than reduced to one aggregate score.
- A controlled prompt edit demonstrates both an improvement and a regression, with a rollback path.

**Starting material:** `prompt-regression`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
