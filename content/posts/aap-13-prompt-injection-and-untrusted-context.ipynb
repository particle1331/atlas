This draft describes the lesson's scope and the implementation it will examine. The companion code and measured results are not available yet.

## Topics

- Trusted instructions versus retrieved/supplied content; intended task and attacker goal.
- No defence, literal filtering, model classification and prompt separation.
- Paraphrases, delimiter imitation and benign quoted instructions.
- Attack success and false rejection; structured verdict handling and external permissions.

## Project code

**Planned companion code:** `projects/applied-ai-prototypes/aap13` (relative to the repository root).

- A document-QA fixture separates the intended task from attacker-controlled retrieved text.
- Defence adapters implement literal filtering, model classification, and prompt separation beside an independently constrained action interface.
- A mixed benign/adversarial corpus measures false rejection and attack success, including paraphrases and delimiter imitation.

**Starting material:** `prompt-injection`.

The source examples are starting points, not verified results for this draft. Our implementation will be maintained separately from the upstream reference revision.
