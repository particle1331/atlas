This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Trusted instructions versus retrieved/supplied content; intended task and attacker goal.
- No defence, literal filtering, model classification and prompt separation.
- Paraphrases, delimiter imitation and benign quoted instructions.
- Attack success and false rejection; structured verdict handling and external permissions.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap13` (relative to the repository root).

**Imported reference examples:** `reference/prompt-injection/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A document-QA fixture separates the intended task from attacker-controlled retrieved text.
- Defence adapters implement literal filtering, model classification, and prompt separation beside an independently constrained action interface.
- A mixed benign/adversarial corpus measures false rejection and attack success, including paraphrases and delimiter imitation.

**Starting material:** `prompt-injection`.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
