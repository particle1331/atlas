This draft describes the lesson's scope and the implementation it will examine. Pinned upstream reference examples are available in the companion project; lesson-specific adaptations and measured results have not been developed.

## Topics

- Task specification, representative sampling, stratification, provenance and held-out cases.
- Exact/domain checks, retrieval metrics, task success, rubric/semantic judges and abstention.
- Judge disagreement, invalid outputs, uncertainty and evaluation cost.
- Development, regression, release and sampled online integration; exploratory failures become reproducible cases.

## Project code

**Companion project:** `projects/applied-ai-prototypes/aap38` (relative to the repository root).

**Imported reference examples:** `reference/evals/`, `reference/prompt-regression/`, `reference/rag-failure-modes/`.

These are reference copies, not a finished lesson implementation. The project README records provenance, execution precautions and any missing parts. No prototype was run during import.

- A versioned dataset/fixture registry and evaluation runner separate exact, domain, retrieval, task, and model-judge checks.
- Scoring adapters validate judge output, abstention, disagreement, and usage while retaining per-case evidence.
- Integration examples distinguish development smoke checks, regression suites, release gates, and sampled online review.

**Starting material:** `evals`, `prompt-regression`, metric cases from AAP-21 and adversarial experiments from AAP-6–7/AAP-13–15.

Imported examples remain unchanged under `reference/`; later authored adaptations belong in a separately named directory. The project README and provenance manifest distinguish imported material from pending work.
